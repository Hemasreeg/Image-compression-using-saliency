## 📂 Step 4: Setup Dataset from Google Drive

**Prerequisites:**
1. Download DUTS dataset: http://saliencydetection.net/duts/
2. Upload `DUTS-TR.zip` and `DUTS-TE.zip` to your Google Drive (MyDrive folder)
3. Run the cell below to mount and extract

````python
from google.colab import drive
import zipfile
import os
import glob

print("🔗 Mounting Google Drive...")
drive.mount('/content/drive')

# Extract datasets
drive_path = '/content/drive/MyDrive/'
os.makedirs('DUTS', exist_ok=True)

for dataset in ['DUTS-TR', 'DUTS-TE']:
    zip_file = os.path.join(drive_path, f'{dataset}.zip')
    if os.path.exists(zip_file):
        print(f"📦 Extracting {dataset}...")
        with zipfile.ZipFile(zip_file, 'r') as zip_ref:
            zip_ref.extractall('DUTS')
        print(f"✓ {dataset} extracted")
    else:
        print(f"❌ {dataset}.zip not found in Google Drive")

# Verify
train_imgs = len(glob.glob('DUTS/DUTS-TR/DUTS-TR-Image/*.jpg'))
test_imgs = len(glob.glob('DUTS/DUTS-TE/DUTS-TE-Image/*.jpg'))
print(f"\n✓ Train images: {train_imgs}")
print(f"✓ Test images: {test_imgs}")

if train_imgs > 0 and test_imgs > 0:
    print("\n✅ Dataset ready for training!")
else:
    print("\n❌ Dataset extraction failed - check folder structure")
````

## 🏋️ Step 6: Train Enhanced Model (20 epochs)

**Training Strategy:**
- Epochs: 20
- Batch size: 8 (Colab free tier GPU)
- Base layers: Start from pre-trained, fine-tune during training
- Novel components: Train from scratch (random initialization)
- Output: `enhanced_u2net.pth` with TRAINED novel components

**Estimated time: 2-4 hours on Tesla T4**

````python
# Pre-training validation
import os

errors = []
if not os.path.exists('train_enhanced.py'):
    errors.append("train_enhanced.py missing - Go to Step 3")
if not os.path.exists('DUTS/DUTS-TR'):
    errors.append("DUTS dataset missing - Complete Step 4")
if not os.path.exists('models/best_model.pth'):
    errors.append("Pre-trained model missing - Complete Step 5")

if errors:
    print("❌ Cannot start training:\n")
    for err in errors:
        print(f"  • {err}")
    raise SystemExit("Fix errors above before training")

print("✅ All prerequisites met. Starting training...\n")
print("="*70)
print("  TRAINING ENHANCED U2-NET WITH NOVEL COMPONENTS")
print("="*70)
print("\n📊 Model Architecture:")
print("  • Base U2-Net: 44,009,797 params (from pre-trained)")
print("  • Novel Components: 777,612 params (train from scratch)")
print("  • Total: 44,787,409 params\n")
print("🎯 Novel Components Training:")
print("  • ChannelAttention (bottleneck)")
print("  • EdgeAwareRefinement (4 instances)")
print("  • MultiScaleAdaptiveFusion")
print("  • SpatialAttention")
print("="*70)
print()

from train_enhanced import train_enhanced_model

train_enhanced_model(
    data_dir='DUTS',
    output_dir='models',
    epochs=20,
    batch_size=8,
    learning_rate=0.0001,
    pretrained_path='models/best_model.pth'
)

print("\n✅ Training complete!")
print("\n📦 Output files:")
print("  • models/enhanced_u2net.pth - Best model with trained novel components")
print("  • models/training_history.json - All metrics")
print("  • models/checkpoint_epoch_*.pth - Checkpoints every 5 epochs")
````

## 📊 Step 7: Plot Training Results

````python
import json
import matplotlib.pyplot as plt

# Load training history
with open('models/training_history.json', 'r') as f:
    history = json.load(f)

fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# F1 Score
axes[0].plot(history['epochs'], history['train_f1'], 'b-', label='Train F1', linewidth=2)
axes[0].plot(history['epochs'], history['val_f1'], 'r-', label='Val F1', linewidth=2)
axes[0].set_xlabel('Epoch', fontsize=12)
axes[0].set_ylabel('F1 Score', fontsize=12)
axes[0].set_title('F1 Score Progress (Novel Components Training)', fontsize=14, fontweight='bold')
axes[0].legend(fontsize=10)
axes[0].grid(True, alpha=0.3)

# Loss
axes[1].plot(history['epochs'], history['train_loss'], 'b-', label='Train Loss', linewidth=2)
axes[1].plot(history['epochs'], history['val_loss'], 'r-', label='Val Loss', linewidth=2)
axes[1].set_xlabel('Epoch', fontsize=12)
axes[1].set_ylabel('Loss', fontsize=12)
axes[1].set_title('Loss Progress', fontsize=14, fontweight='bold')
axes[1].legend(fontsize=10)
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('training_results.png', dpi=300, bbox_inches='tight')
plt.show()

print(f"\n📊 Final Results:")
print(f"  Best Val F1: {max(history['val_f1']):.4f}")
print(f"  Final Train Loss: {history['train_loss'][-1]:.4f}")
print(f"  Final Val Loss: {history['val_loss'][-1]:.4f}")
print("\n✅ Use these metrics to compare with base U2-Net!")
````
