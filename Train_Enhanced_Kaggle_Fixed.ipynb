## 🔍 Step 6: Verify Enhanced Model Architecture

```python
import torch
from model import U2NET

print("="*70)
print("  VERIFYING ENHANCED U2-NET ARCHITECTURE")
print("="*70)

model = U2NET(3, 1)
total_params = sum(p.numel() for p in model.parameters())

# Count novel component parameters with correct attribute names
novel_params = 0
novel_components = []

# Check bottleneck channel attention
if hasattr(model, 'bottleneck_ca'):
    ca_params = sum(p.numel() for p in model.bottleneck_ca.parameters())
    novel_params += ca_params
    novel_components.append(f"✓ ChannelAttention (bottleneck): {ca_params:,} params")

# Check edge refinement modules
for i in range(1, 5):
    edge_refine = getattr(model, f'edge_refine_{i}', None)
    if edge_refine:
        er_params = sum(p.numel() for p in edge_refine.parameters())
        novel_params += er_params
        novel_components.append(f"✓ EdgeAwareRefinement{i}: {er_params:,} params")

# Check adaptive fusion
if hasattr(model, 'adaptive_fusion'):
    af_params = sum(p.numel() for p in model.adaptive_fusion.parameters())
    novel_params += af_params
    novel_components.append(f"✓ MultiScaleAdaptiveFusion: {af_params:,} params")

print(f"\n📊 Total Parameters: {total_params:,}")
print(f"📊 Base U2-Net: {total_params - novel_params:,} ({(total_params-novel_params)/total_params*100:.2f}%)")
print(f"📊 Novel Components: {novel_params:,} ({novel_params/total_params*100:.2f}%)\n")

if novel_components:
    print("🎯 Novel Components Detected:")
    for comp in novel_components:
        print(f"  {comp}")
else:
    print("❌ No novel components found! Check model.py")

# Test forward pass
dummy_input = torch.randn(2, 3, 320, 320)
output = model(dummy_input)
print(f"\n✓ Forward pass successful: {dummy_input.shape} → {output[0].shape}")
print(f"✓ Output includes {len(output)} scales (d0-d6)")
print("\n✅ Enhanced architecture verified and ready for training!")
```

```python
# 🚀 DenseCRF Post-processing for Ensemble Output
# Install pydensecrf if not present (Kaggle: uncomment next line if needed)
# !pip install pydensecrf
import numpy as np
import cv2
from ensemble_model import EnsembleModel
from PIL import Image
import matplotlib.pyplot as plt
try:
    import pydensecrf.densecrf as dcrf
    from pydensecrf.utils import unary_from_softmax
except ImportError:
    print('Please install pydensecrf for DenseCRF post-processing.')

def apply_densecrf(image, mask, n_iters=5):
    h, w = mask.shape
    d = dcrf.DenseCRF2D(w, h, 2)
    # Prepare softmax: 2 classes (foreground/background)
    softmax = np.zeros((2, h, w), dtype=np.float32)
    softmax[1] = mask
    softmax[0] = 1 - mask
    unary = unary_from_softmax(softmax)
    d.setUnaryEnergy(unary)
    d.addPairwiseGaussian(sxy=3, compat=3)
    d.addPairwiseBilateral(sxy=80, srgb=13, rgbim=np.array(image), compat=10)
    Q = d.inference(n_iters)
    refined = np.array(Q[1]).reshape((h, w))
    return refined

# Example usage:
model_paths = sorted(glob.glob('models/*.pth'))
if not model_paths:
    print('❌ No model checkpoints found in models/. Please train or upload .pth files.')
else:
    best_model = [p for p in model_paths if 'best' in p.lower()]
    u2net_path = best_model[0] if best_model else model_paths[0]
    ensemble = EnsembleModel(u2net_path=u2net_path)
    test_image_path = 'DUTS/DUTS-TE/DUTS-TE-Image/your_test_image.jpg'  # <-- Change to your test image
    image = Image.open(test_image_path).convert('RGB')
    mask = ensemble.predict(image)
    # Apply DenseCRF
    refined_mask = apply_densecrf(np.array(image), mask)
    # Visualize
    plt.figure(figsize=(18,5))
    plt.subplot(1,3,1)
    plt.imshow(image)
    plt.title('Input Image')
    plt.axis('off')
    plt.subplot(1,3,2)
    plt.imshow(mask, cmap='gray')
    plt.title('Ensemble Mask')
    plt.axis('off')
    plt.subplot(1,3,3)
    plt.imshow(refined_mask, cmap='gray')
    plt.title('DenseCRF Refined Mask')
    plt.axis('off')
    plt.show()
```
